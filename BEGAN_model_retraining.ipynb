```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset
import torchvision.transforms as transforms
import torchvision.utils as vutils
from torch.cuda.amp import autocast, GradScaler # TĂNG TỐC ĐỘ

import os
import zipfile
import shutil
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import drive

# ==========================================
# 1. CẤU HÌNH HỆ THỐNG
# ==========================================
# Mount Google Drive
drive.mount('/content/drive')

# --- CẤU HÌNH ĐƯỜNG DẪN (SỬA NẾU KHÁC) ---
PROJECT_DIR = "/content/drive/MyDrive/BEGAN_PROJECT"
ZIP_FILE_PATH = os.path.join(PROJECT_DIR, "images.zip") # File zip ảnh gốc
CHECKPOINT_DIR = os.path.join(PROJECT_DIR, "began_checkpoints")
SAMPLE_DIR = os.path.join(PROJECT_DIR, "began_samples")

# Tạo thư mục nếu chưa có
os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(SAMPLE_DIR, exist_ok=True)

# --- HYPERPARAMETERS ---
IMAGE_SIZE = 128
Z_DIM = 64
BATCH_SIZE = 256    # A100 có thể lên 256. T4 để 64-128.
NUM_WORKERS = 4     # Số luồng CPU đọc dữ liệu
LR = 0.0001
GAMMA = 0.5         # BEGAN equilibrium gamma
LAMBDA_K = 0.001    # Tốc độ cập nhật k
EPOCHS = 200        # Train đủ lâu để hết "đốm màu"

DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'

# Tối ưu hóa CUDNN
torch.backends.cudnn.benchmark = True

# ==========================================
# 2. CHUẨN BỊ DỮ LIỆU (TỐI ƯU TỐC ĐỘ)
# ==========================================
LOCAL_DATA_DIR = "/content/dataset"

def prepare_data():
    if not os.path.exists(LOCAL_DATA_DIR):
        print(f"--> Đang giải nén dữ liệu từ Drive xuống {LOCAL_DATA_DIR} để tăng tốc...")
        try:
            with zipfile.ZipFile(ZIP_FILE_PATH, 'r') as zip_ref:
                zip_ref.extractall(LOCAL_DATA_DIR)
            print("✅ Giải nén hoàn tất!")
        except Exception as e:
            print(f"❌ Lỗi giải nén: {e}")
            return False
    else:
        print("✅ Dữ liệu đã có sẵn ở máy ảo.")
    return True

class FaceDataset(Dataset):
    def __init__(self, root_dir):
        # Tìm tất cả file ảnh trong thư mục (bao gồm thư mục con)
        self.image_paths = []
        for root, _, files in os.walk(root_dir):
            for file in files:
                if file.lower().endswith(('png', 'jpg', 'jpeg')):
                    self.image_paths.append(os.path.join(root, file))

        print(f"--> Đã tìm thấy {len(self.image_paths)} ảnh.")

        self.transform = transforms.Compose([
            transforms.ToPILImage(),
            transforms.Resize((IMAGE_SIZE, IMAGE_SIZE)),
            transforms.ToTensor(),
            transforms.Normalize((0.5, 0.5, 0.5), (0.5, 0.5, 0.5))
        ])

    def __len__(self):
        return len(self.image_paths)

    def __getitem__(self, idx):
        img_path = self.image_paths[idx]
        try:
            img = cv2.imread(img_path)
            img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
            return self.transform(img)
        except:
            # Fallback nếu ảnh lỗi
            return torch.zeros((3, IMAGE_SIZE, IMAGE_SIZE))

# ==========================================
# 3. KIẾN TRÚC MÔ HÌNH (BEGAN 128x128)
# ==========================================
class Decoder(nn.Module):
    def __init__(self):
        super(Decoder, self).__init__()
        self.fc = nn.Linear(Z_DIM, 8 * 8 * 64)
        self.layers = nn.Sequential(
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.Upsample(scale_factor=2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.Upsample(scale_factor=2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.Upsample(scale_factor=2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.Upsample(scale_factor=2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(),
            nn.Conv2d(64, 3, 3, 1, 1), nn.Tanh()
        )
    def forward(self, x):
        x = self.fc(x)
        x = x.view(-1, 64, 8, 8)
        return self.layers(x)

class Encoder(nn.Module):
    def __init__(self):
        super(Encoder, self).__init__()
        self.layers = nn.Sequential(
            nn.Conv2d(3, 64, 3, 1, 1), nn.ELU(),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.AvgPool2d(2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.AvgPool2d(2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.AvgPool2d(2),
            nn.Conv2d(64, 64, 3, 1, 1), nn.ELU(), nn.AvgPool2d(2),
        )
        self.fc = nn.Linear(8 * 8 * 64, Z_DIM)
    def forward(self, x):
        x = self.layers(x)
        x = x.view(x.size(0), -1)
        return self.fc(x)

class Discriminator(nn.Module):
    def __init__(self):
        super(Discriminator, self).__init__()
        self.encoder = Encoder()
        self.decoder = Decoder()
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ==========================================
# 4. QUẢN LÝ CHECKPOINT (QUAN TRỌNG)
# ==========================================
def save_checkpoint(epoch, generator, discriminator, opt_g, opt_d, k, measure, filename="latest_checkpoint.pth"):
    save_path = os.path.join(CHECKPOINT_DIR, filename)
    torch.save({
        'epoch': epoch,
        'generator_state_dict': generator.state_dict(),
        'discriminator_state_dict': discriminator.state_dict(),
        'optimizer_G_state_dict': opt_g.state_dict(),
        'optimizer_D_state_dict': opt_d.state_dict(),
        'k': k,
        'measure': measure
    }, save_path)
    print(f"--> [Safe Save] Đã lưu checkpoint: {filename}")

def load_checkpoint(generator, discriminator, opt_g, opt_d):
    # Ưu tiên load file 'latest'
    latest_path = os.path.join(CHECKPOINT_DIR, "latest_checkpoint.pth")
    if os.path.exists(latest_path):
        print(f"--> Phát hiện checkpoint cũ. Đang khôi phục từ: {latest_path}")
        checkpoint = torch.load(latest_path, map_location=DEVICE)

        generator.load_state_dict(checkpoint['generator_state_dict'])
        discriminator.load_state_dict(checkpoint['discriminator_state_dict'])
        opt_g.load_state_dict(checkpoint['optimizer_G_state_dict'])
        opt_d.load_state_dict(checkpoint['optimizer_D_state_dict'])

        start_epoch = checkpoint['epoch'] + 1
        k = checkpoint['k']
        measure = checkpoint['measure']
        print(f"✅ Khôi phục thành công! Tiếp tục tại Epoch {start_epoch}")
        return start_epoch, k, measure
    else:
        print("--> Không tìm thấy checkpoint. Bắt đầu train mới.")
        return 1, 0.0, 0.0

# ==========================================
# 5. VÒNG LẶP TRAINING (MAIN)
# ==========================================
def train():
    if not prepare_data(): return

    # Dataset
    dataset = FaceDataset(LOCAL_DATA_DIR)
    dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True,
                            num_workers=NUM_WORKERS, pin_memory=True)

    # Models
    generator = Decoder().to(DEVICE)
    discriminator = Discriminator().to(DEVICE)

    # Optimizers
    opt_g = optim.Adam(generator.parameters(), lr=LR)
    opt_d = optim.Adam(discriminator.parameters(), lr=LR)

    # Khôi phục nếu có sự cố
    start_epoch, k, measure = load_checkpoint(generator, discriminator, opt_g, opt_d)

    # AMP Scaler
    scaler = GradScaler()

    # Cố định z để theo dõi tiến độ sinh ảnh
    fixed_z = torch.randn(25, Z_DIM).to(DEVICE)

    print(f"--- BẮT ĐẦU TRAINING BEGAN (AMP Enabled) ---")

    for epoch in range(start_epoch, EPOCHS + 1):
        g_loss_accum = 0.0
        d_loss_accum = 0.0

        for i, real_imgs in enumerate(dataloader):
            real_imgs = real_imgs.to(DEVICE)
            bs = real_imgs.size(0)

            # --- TRAIN GENERATOR ---
            opt_g.zero_grad()
            z = torch.randn(bs, Z_DIM).to(DEVICE)

            with autocast(): # Tự động chuyển float32 -> float16
                gen_imgs = generator(z)
                recon_gen = discriminator(gen_imgs)
                g_loss = torch.mean(torch.abs(gen_imgs - recon_gen)) # L1 Loss

            scaler.scale(g_loss).backward()
            scaler.step(opt_g)
            scaler.update()

            # --- TRAIN DISCRIMINATOR ---
            opt_d.zero_grad()

            with autocast():
                # Tính lại recon_gen để không cần detach graph phức tạp (hoặc dùng gen_imgs.detach())
                recon_real = discriminator(real_imgs)
                recon_gen = discriminator(gen_imgs.detach())

                d_loss_real = torch.mean(torch.abs(real_imgs - recon_real))
                d_loss_fake = torch.mean(torch.abs(gen_imgs.detach() - recon_gen))
                d_loss = d_loss_real - k * d_loss_fake

            scaler.scale(d_loss).backward()
            scaler.step(opt_d)
            scaler.update()

            # --- UPDATE k & MEASURE ---
            with torch.no_grad():
                balance = (GAMMA * d_loss_real.item()) - d_loss_fake.item()
                k = k + LAMBDA_K * balance
                k = max(min(k, 1.0), 0.0) # Kẹp k trong [0, 1]
                measure = d_loss_real.item() + abs(balance)

            g_loss_accum += g_loss.item()
            d_loss_accum += d_loss.item()

            if i % 100 == 0:
                print(f"[Epoch {epoch}][Batch {i}] G_Loss: {g_loss.item():.4f} | D_Loss: {d_loss.item():.4f} | M: {measure:.4f} | k: {k:.4f}")

        # --- CUỐI MỖI EPOCH ---

        # 1. Lưu ảnh mẫu 5x5
        with torch.no_grad():
            sample_imgs = generator(fixed_z)
            vutils.save_image(sample_imgs,
                              os.path.join(SAMPLE_DIR, f"epoch_{epoch}.png"),
                              nrow=5, normalize=True)
            # In ra màn hình Colab để xem luôn
            if epoch % 5 == 0: # 5 epoch in 1 lần cho đỡ lag
                grid = vutils.make_grid(sample_imgs[:5], nrow=5, normalize=True)
                plt.imshow(grid.cpu().permute(1,2,0).numpy())
                plt.axis('off')
                plt.title(f"Epoch {epoch}")
                plt.show()

        # 2. Lưu checkpoint an toàn (Overwrite 'latest' và lưu riêng file epoch)
        # Lưu file 'latest' để resume
        save_checkpoint(epoch, generator, discriminator, opt_g, opt_d, k, measure, "latest_checkpoint.pth")

        # Cứ 10 epoch thì backup 1 file riêng để tránh file latest bị hỏng
        if epoch % 10 == 0:
            save_checkpoint(epoch, generator, discriminator, opt_g, opt_d, k, measure, f"checkpoint_epoch_{epoch}.pth")

        # 3. Lưu model weights riêng (Dễ dùng cho bước sau)
        if epoch % 5 == 0:
            torch.save(generator.state_dict(), os.path.join(CHECKPOINT_DIR, f"generator_epoch_{epoch}.pth"))

if __name__ == "__main__":
    train()
```

```python
import torch
import torchvision.utils as vutils
import matplotlib.pyplot as plt
import numpy as np
import os

# --- CẤU HÌNH ---
DEVICE = 'cuda' if torch.cuda.is_available() else 'cpu'
Z_DIM = 64
MODEL_PATH = "/content/drive/MyDrive/BEGAN_PROJECT/began_checkpoints/generator_epoch_100.pth" # Sửa đường dẫn nếu cần

# --- LOAD MODEL ---
# (Đảm bảo class Decoder đã được định nghĩa trong notebook của bạn)
generator = Decoder().to(DEVICE)
try:
    generator.load_state_dict(torch.load(MODEL_PATH, map_location=DEVICE))
except:
    # Trường hợp file lưu cả checkpoint, chỉ lấy weight của generator
    checkpoint = torch.load(MODEL_PATH, map_location=DEVICE)
    if 'generator_state_dict' in checkpoint:
        generator.load_state_dict(checkpoint['generator_state_dict'])
    else:
        generator.load_state_dict(checkpoint)

generator.eval()
print("✅ Đã load model thành công!")

# --- TEST 1: KIỂM TRA ĐA DẠNG (Sinh 32 ảnh) ---
def check_diversity():
    print("\n--- BÀI 1: KIỂM TRA ĐA DẠNG ---")
    with torch.no_grad():
        z = torch.randn(32, Z_DIM).to(DEVICE)
        fake_imgs = generator(z)

        # Denormalize (-1,1) -> (0,1)
        fake_imgs = (fake_imgs + 1) / 2.0

        grid = vutils.make_grid(fake_imgs, nrow=8, padding=2)
        plt.figure(figsize=(15, 8))
        plt.axis("off")
        plt.title("Đa dạng mẫu sinh ra tại Epoch 100")
        plt.imshow(grid.cpu().permute(1, 2, 0))
        plt.show()

# --- TEST 2: KIỂM TRA NỘI SUY (Biến hình) ---
def check_interpolation():
    print("\n--- BÀI 2: KIỂM TRA NỘI SUY (INTERPOLATION) ---")
    # Tạo 2 vector ngẫu nhiên
    z1 = torch.randn(1, Z_DIM).to(DEVICE)
    z2 = torch.randn(1, Z_DIM).to(DEVICE)

    # Tạo 10 bước chuyển đổi
    alphas = np.linspace(0, 1, 10)
    z_interp = []
    for alpha in alphas:
        z_curr = (1 - alpha) * z1 + alpha * z2
        z_interp.append(z_curr)

    z_interp = torch.cat(z_interp, dim=0)

    with torch.no_grad():
        imgs = generator(z_interp)
        imgs = (imgs + 1) / 2.0

        grid = vutils.make_grid(imgs, nrow=10, padding=2)
        plt.figure(figsize=(15, 3))
        plt.axis("off")
        plt.title("Biến đổi từ Người A -> Người B")
        plt.imshow(grid.cpu().permute(1, 2, 0))
        plt.show()

# --- THỰC THI ---
check_diversity()
check_interpolation()
```
